```python
import pandas as pd
```

```python
def row_to_text(row):
    return (
        f"Product: {row['Product']}. "
        f"Price: {row['Price']}. "
        f"Category: {row['Category']}. "
        f"Description: {row['Description']}"
    )
```

```python
df = pd.read_csv("data/boutique_ecommerce_100_products.csv")
df["text"] = df.apply(row_to_text, axis=1)
```
